```python
import seaborn as sns
import pandas as pd
import numpy as np
import sys
import matplotlib as mb
import matplotlib.pyplot as plt
import os
```

```python
# Making the python code executable on the first argument
if __name__ == "__main__":
    if len(sys.argv) < 2:  
        print("Usage: python csv_to_dataframe.py <csv_file>")
        sys.exit(1)

    csv_file = sys.argv[1]  
```

```python
#Extract the filename with and without extension
csv_filename = os.path.basename(csv_file)
csv_title = os.path.splitext(csv_filename)[0]  # Get filename without extension
```

```python
#Reading the csv file into a pandas dataframe
df = pd.read_csv(csv_file)
```

```python
fura_columns = df.filter(regex='^Fura \d$')
```

```python
df['Fura Average'] = fura_columns.mean(axis=1)
```

```python
sns.boxplot(data=fura_columns , width=0.5, fliersize=3)
plt.title(f'Data from {csv_title}')
plt.savefig('OutlierCheck.png') 
plt.show()
```
