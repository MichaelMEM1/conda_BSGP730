```python
import seaborn as sns
import pandas as pd
import matplotlib as mb
import matplotlib.pyplot as plt
import numpy as np
import scipy as sc
```

```python
ankoff_df = pd.read_csv('CakcTranDataAnkOff.csv')
```

```python
print(ankoff_df.head())
```

```python
# Selecting only the Fura columns not average
fura_columns = ankoff_df.filter(regex='^Fura \d$')
```

```python
ankoff_df['Fura Average'] = fura_columns.mean(axis=1)
```

```python
print(ankoff_df.head())
```

```python
ankoff_df.rename(columns={"Time ":"Time"}, inplace=True)
```

```python
sns.lineplot(x="Time", y="Fura Average",
             data=ankoff_df)
plt.title('Average Calcium Transients')
plt.show()
```

```python
sns.boxplot(data=fura_columns , width=0.5, fliersize=3)
plt.title("Boxplot for All Fura Columns")
plt.show()
```
