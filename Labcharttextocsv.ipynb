```python
import csv
```

```python

# Define the input and output file names
input_filename = "ank ATRAIL 10 Seconds 7-29-2024 0.5 SD.txt"

# Clean up the input filename to create a suitable CSV output filename
output_filename = "ECGank.csv"

# Open the input text file and the output CSV file
with open(input_filename, 'r') as infile, open(output_filename, 'w', newline='') as outfile:
    stripped = (line.strip() for line in infile)  # Strip leading/trailing whitespace
    header_lines = []  # List to store header lines
    data_lines = []  # List to store data lines

    # Manually define column names
    column_names = ['Time', 'Channel 1', 'Channel 2', 'Channel 3']  # Example column names

    for line in stripped:
        if "=" in line:  # Check if the line contains a key-value pair (header line)
            header_lines.append(line)  # Store header line
            
        elif line:  # Process non-empty lines (which are data lines)
            data_lines.append(line.split())  # Split by whitespace

    # Write column names as the first row (manual column names)
    writer = csv.writer(outfile)
    writer.writerow(column_names)  # Write the header row

    # Write data rows
    writer.writerows(data_lines)

print(f"CSV file saved as: {output_filename}")
```
